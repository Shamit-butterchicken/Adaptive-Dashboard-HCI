```python
# ============================================================
# 1. IMPORTS & ENVIRONMENT SETUP
# ============================================================
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets

import tensorflow as tf
from tensorflow.keras import layers, models, regularizers, metrics, optimizers, callbacks
import tensorflow_probability as tfp

# Ensure output directories exist
os.makedirs("./output", exist_ok=True)
os.makedirs("./logs", exist_ok=True)

# ============================================================
# 2. HYPERPARAMETERS
# ============================================================
COUPLING_DIM = 256
COUPLING_LAYERS = 2
INPUT_DIM = 2
REGULARIZATION = 0.01
BATCH_SIZE = 256
EPOCHS = 600

# ============================================================
# 3. DATA PREPARATION
# ============================================================
data = datasets.make_moons(30000, noise=0.05)[0].astype("float32")
norm = layers.Normalization()
norm.adapt(data)
normalized_data = norm(data).numpy()

# ============================================================
# 4. COUPLING LAYER ARCHITECTURE (Fixed Input shape)
# ============================================================
def Coupling(input_dim, coupling_dim, reg):
    # Fixed: shape must be a tuple (input_dim,)
    input_layer = layers.Input(shape=(input_dim,))

    # Scale network (s) -> output activation 'tanh'
    s_layer = layers.Dense(coupling_dim, activation="relu", kernel_regularizer=regularizers.l2(reg))(input_layer)
    s_layer = layers.Dense(coupling_dim, activation="relu", kernel_regularizer=regularizers.l2(reg))(s_layer)
    s_layer = layers.Dense(coupling_dim, activation="relu", kernel_regularizer=regularizers.l2(reg))(s_layer)
    s_layer = layers.Dense(coupling_dim, activation="relu", kernel_regularizer=regularizers.l2(reg))(s_layer)
    s_layer_out = layers.Dense(input_dim, activation="tanh", kernel_regularizer=regularizers.l2(reg))(s_layer)

    # Translation network (t) -> output activation 'linear'
    t_layer = layers.Dense(coupling_dim, activation="relu", kernel_regularizer=regularizers.l2(reg))(input_layer)
    t_layer = layers.Dense(coupling_dim, activation="relu", kernel_regularizer=regularizers.l2(reg))(t_layer)
    t_layer = layers.Dense(coupling_dim, activation="relu", kernel_regularizer=regularizers.l2(reg))(t_layer)
    t_layer = layers.Dense(coupling_dim, activation="relu", kernel_regularizer=regularizers.l2(reg))(t_layer)
    t_layer_out = layers.Dense(input_dim, activation="linear", kernel_regularizer=regularizers.l2(reg))(t_layer)

    return models.Model(inputs=input_layer, outputs=[s_layer_out, t_layer_out])

# ============================================================
# 5. REALNVP MODEL CLASS
# ============================================================
class RealNVP(models.Model):
    def __init__(self, input_dim, coupling_layers, coupling_dim, regularization):
        super().__init__()
        self.coupling_layers = coupling_layers
        self.distribution = tfp.distributions.MultivariateNormalDiag(
            loc=[0.0, 0.0], scale_diag=[1.0, 1.0]
        )
        self.masks = np.array(
            [[0, 1], [1, 0]] * (coupling_layers // 2), dtype="float32"
        )
        self.loss_tracker = metrics.Mean(name="loss")
        self.layers_list = [
            Coupling(input_dim, coupling_dim, regularization)
            for i in range(coupling_layers)
        ]

    @property
    def metrics(self):
        return [self.loss_tracker]

    def call(self, x, training=True):
        log_det_inv = 0
        direction = 1
        if training:
            direction = -1
        for i in range(self.coupling_layers)[::direction]:
            x_masked = x * self.masks[i]
            reversed_mask = 1 - self.masks[i]
            s, t = self.layers_list[i](x_masked)
            s *= reversed_mask
            t *= reversed_mask
            gate = (direction - 1) / 2
            x = (
                reversed_mask
                * (x * tf.exp(direction * s) + direction * t * tf.exp(gate * s))
                + x_masked
            )
            log_det_inv += gate * tf.reduce_sum(s, axis=1)
        return x, log_det_inv

    def log_loss(self, x):
        y, logdet = self(x)
        log_likelihood = self.distribution.log_prob(y) + logdet
        return -tf.reduce_mean(log_likelihood)

    def train_step(self, data):
        with tf.GradientTape() as tape:
            loss = self.log_loss(data)
        g = tape.gradient(loss, self.trainable_variables)
        self.optimizer.apply_gradients(zip(g, self.trainable_variables))
        self.loss_tracker.update_state(loss)
        return {"loss": self.loss_tracker.result()}

    def test_step(self, data):
        loss = self.log_loss(data)
        self.loss_tracker.update_state(loss)
        return {"loss": self.loss_tracker.result()}

model = RealNVP(
    input_dim=INPUT_DIM,
    coupling_layers=COUPLING_LAYERS,
    coupling_dim=COUPLING_DIM,
    regularization=REGULARIZATION,
)
model.compile(optimizer=optimizers.Adam(learning_rate=0.0001))

# ============================================================
# 6. IMAGE GENERATOR CALLBACK
# ============================================================
class ImageGenerator(callbacks.Callback):
    def __init__(self, num_samples):
        super().__init__()
        self.num_samples = num_samples

    def generate(self):
        z, _ = self.model(normalized_data, training=True)
        samples = self.model.distribution.sample(self.num_samples)
        x, _ = self.model(samples, training=False)
        return np.array(x), np.array(z), np.array(samples)

    def display(self, x, z, samples, save_to=None):
        f, axes = plt.subplots(2, 2, figsize=(8, 6))

        axes[0, 0].scatter(normalized_data[:, 0], normalized_data[:, 1], color="r", s=1)
        axes[0, 0].set(title="Data space X", xlabel="x_1", ylabel="x_2")
        axes[0, 0].set_xlim([-2, 2])
        axes[0, 0].set_ylim([-2, 2])

        axes[0, 1].scatter(z[:, 0], z[:, 1], color="r", s=1)
        axes[0, 1].set(title="Forward Transform f(X)", xlabel="z_1", ylabel="z_2")
        axes[0, 1].set_xlim([-2, 2])
        axes[0, 1].set_ylim([-2, 2])

        axes[1, 0].scatter(samples[:, 0], samples[:, 1], color="g", s=1)
        axes[1, 0].set(title="Latent space Z (Gaussian)", xlabel="z_1", ylabel="z_2")
        axes[1, 0].set_xlim([-2, 2])
        axes[1, 0].set_ylim([-2, 2])

        axes[1, 1].scatter(x[:, 0], x[:, 1], color="g", s=1)
        axes[1, 1].set(title="Inverse Transform g(Z)", xlabel="x_1", ylabel="x_2")
        axes[1, 1].set_xlim([-2, 2])
        axes[1, 1].set_ylim([-2, 2])

        plt.subplots_adjust(wspace=0.3, hspace=0.5)
        if save_to:
            plt.savefig(save_to)
        plt.show()

    def on_epoch_end(self, epoch, logs=None):
        if (epoch + 1) % 100 == 0 or epoch == EPOCHS - 1:
            x, z, samples = self.generate()
            self.display(x, z, samples, save_to=f"./output/generated_img_{epoch:03d}.png")

img_generator_callback = ImageGenerator(num_samples=3000)
tensorboard_callback = callbacks.TensorBoard(log_dir="./logs")

# ============================================================
# 7. MODEL TRAINING
# ============================================================
print(f"Training RealNVP for {EPOCHS} epochs...")
history = model.fit(
    normalized_data,
    batch_size=BATCH_SIZE,
    epochs=EPOCHS,
    callbacks=[tensorboard_callback, img_generator_callback],
    verbose=1,
)

# ============================================================
# 8. GENERATE & SAVE LOSS CURVE
# ============================================================
plt.figure(figsize=(9, 5))
plt.plot(range(1, EPOCHS + 1), history.history["loss"], color="#1f77b4", linewidth=2, label="Negative Log-Likelihood Loss")
plt.axvline(x=300, color="red", linestyle="--", label="Book Baseline (300 Epochs)")
plt.title("RealNVP Training Loss Curve (600 Epochs)", fontsize=14, fontweight="bold")
plt.xlabel("Epoch", fontsize=12)
plt.ylabel("Loss (-Log Likelihood)", fontsize=12)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()
plt.savefig("loss_curve.png", dpi=300)
plt.show()
```
